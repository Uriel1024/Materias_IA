## Modulo para clasificar los nuevos dato de entranada


```python
# 2. Función de validación para los datos nuevos
def evaluar_solo_nuevos(modelo_entrenado, framework='tensorflow'):
    """
    Evalúa el modelo entrenado usando estrictamente las 15 muestras nuevas.
    framework: 'tensorflow' o 'pytorch'
    """
    print(f"\n--- Evaluando 15 DATOS NUEVOS con modelo {framework.upper()} ---")

    predicciones = []

    # Adaptación de la predicción según el framework
    if framework == 'tensorflow':
        X_np = np.array(X_NUEVOS, dtype=np.float32)
        # Obtenemos predicciones crudas
        res_crudas = modelo_entrenado.predict(X_np, verbose=0)
        # argmax para obtener la clase (0, 1 o 2)
        predicciones = np.argmax(res_crudas, axis=1).tolist()

    elif framework == 'pytorch':
        X_tensor = torch.tensor(X_NUEVOS, dtype=torch.float32)
        # Modo evaluación sin cálculo de gradientes
        with torch.no_grad():
            res_crudas = modelo_entrenado(X_tensor)
            predicciones = torch.argmax(res_crudas, dim=1).tolist()

    # Evaluación y conteo de aciertos
    aciertos = 0
    matriz_nuevos = [[0, 0, 0] for _ in range(3)]

    for i in range(len(X_NUEVOS)):
        real = int(Y_NUEVOS[i])
        pred = int(predicciones[i])

        # Poblar la matriz de confusión de estos 15 datos
        matriz_nuevos[real][pred] += 1

        estado = "✅ Correcto" if real == pred else "❌ Incorrecto"
        if real == pred:
            aciertos += 1

        print(f"Muestra {i+1:02d} | Esperado: {real} | Predicción: {pred} -> {estado}")

    precision_porcentaje = (aciertos / len(X_NUEVOS)) * 100
    print(f"\n-> Aciertos en datos nunca antes vistos: {aciertos}/15 ({precision_porcentaje:.2f}%)")

    # Imprimir métricas de la matriz usando la función global que ya tenías
    print("\nMétricas Globales (Solo para los 15 datos nuevos):")
    matrix_confussion(matriz_nuevos, 3)
```

#Modulo para obtener la matriz de confusión.

```python

# 1. Función corregida para obtener la clase predicha
def crear_matriz(res):
    # np.argmax toma las probabilidades de softmax y devuelve
    # el índice de la clase ganadora (0, 1, o 2) para cada fila.
    return np.argmax(res, axis=1).tolist()

# 2. Función corregida para construir la matriz de confusión
def evaluar(res_pred):
    # Inicializar una matriz de 3x3 llena de ceros de manera correcta
    matrix = [[0, 0, 0] for _ in range(3)]

    # Llenar la matriz cruzando la clase real vs la predicha
    for real, pred in zip(CLASS_IRIS, res_pred):
        matrix[int(real)][int(pred)] += 1
    return matrix
```

## Modulo para obtener las métricas de rendimiento del modelo

```python
def matrix_confussion(matrix2, nclases):

  total = sum(matrix2[i][j] for i in range(len(matrix2)) for j in range(len(matrix2[:])) )

  tp = [matrix2[i][i] for i in range(nclases)]
  #los falsos negativos son la suma de las filas de cada clase del y_test restando  los verdaderos positivos
  fn = [(sum(matrix2[i]) - tp[i])for i in range(len(matrix2))]
  #los falsos positivos son la suma de las columnas de cada clase que predijo el modelo restando los verdaderos posivios
  fp = [(sum(col) - tp[i]) for i, col in enumerate(zip(*matrix2))]

  tn = [(total - (tp[i] + fp[i] + fn[i])) for i in range(nclases)]

  specificity = [(tn[i]/(tn[i] + fp[i])) if (tn[i] + fp[i]) != 0 else 0 for i in range(nclases)]
  accuracy = [(tp[i] + tn[i]) / (tp[i] + tn[i] + fp[i] + fn[i]) if (tp[i] + tn[i] + fp[i] + fn[i]) != 0 else 0 for i in range(nclases)]
  precision = [(tp[i]/(tp[i] + fp[i])) if (tp[i] + fp[i]) != 0 else 0 for i in range(nclases) ]
  recall = [(tp[i] / (tp[i] + fn[i]) ) if (tp[i] + fn[i]) != 0 else 0 for i in range(nclases)]

  f1 = [(2*(precision[i]*recall[i])/(precision[i] + recall[i] )) if (precision[i] + recall[i]) != 0 else 0 for i in range(nclases)]

  precisionf = sum(precision) / len(precision)
  recallf = sum(recall) / len(recall)
  f1f = sum(f1) / len(f1)
  accuracyf = sum(accuracy) / len(accuracy)
  specificityf = sum(specificity) / len(specificity)

  accuracy_global = sum(tp) / total
  print("Accuracy global:",accuracy_global)

  print("        Precision      Recall    F1-Score     Specificity     Accuracy")

  for j in range(nclases):
    print(
        f"Clase {j}:    "
        f"{precision[j]:.2f}        "
        f"{recall[j]:.2f}        "
        f"{f1[j]:.2f}        "
        f"{specificity[j]:.2f}        "
        f"{accuracy[j]:.2f}"
    )

  print("-"*90)

  print(
    f"AVG:      "
    f"{precisionf:.2f}        "
    f"{recallf:.2f}        "
    f"{f1f:.2f}        "
    f"{specificityf:.2f}        "
    f"{accuracyf:.2f}"
  )
```

```python
from math import e
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Activation
import random
class Perceptron:
    X = np.empty(0, dtype=np.float32)
    # 2. Salidas esperadas para cada compuerta (Y)
    Y_train = np.empty(0, dtype=np.float32)
    learning_rate = 1.0
    epochs = 100
    verbose = 0
    model = Sequential()
    activation = Activation(None)

    def __init__(self, *args):
        if len(args)==0:
            self.learning_rate = 0.5
            self.epochs = 400
            self.verbose = 0
            #dado que estamos haciendo clasificación multiclase tenemos que usar softmax
            self.activation = Activation('softmax')
        else:
            print(args[0], " ", args[1], " ", args[2], " ", args[3])
            self.learning_rate = float(args[0])
            self.epochs = int(args[1])
            self.verbose = int(args[2])
            self.activation = Activation(args[3])
            #print(self.learning_rate, " ", self.epochs, " ", self.verbose, " ", self.activation.get_config())


    def definirVectorEntrada(self, x):
        self.X = np.array(x,dtype=np.float32)

    def definirSalida(self, entrenamiento):
        self.Y_train = np.array(entrenamiento, dtype=np.float32)

    def definirEpocas(self, epocas):
        self.epochs = epocas

    # 3. Función para crear y entrenar la célula
    def entrenar_celula(self, nombre_compuerta):
        self.compuerta = nombre_compuerta
        print(f"\n--- Entrenando Compuerta {nombre_compuerta} ---")
      # Estructura con 2 entradas (input_shape=[2])
        #cambiamos a input_schape[4] para el vector de 4 caracteristicas que tenemos,
        # units=3 dado que tenemos 3 clases, ocupamos 3 neuronas para clasificar de manera correcta
        self.model = tf.keras.models.Sequential([
            tf.keras.layers.Dense(units=3, input_shape=[4], activation=self.activation)
        ])

        #sparse_categorical_crossentropy para poder hacer clasificación multiclase
        self.model.compile(
            optimizer=tf.keras.optimizers.SGD(learning_rate=self.learning_rate),
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )
        # Entrenar (epochs altas porque es un modelo muy simple con pocos datos)
        print(self.learning_rate, " ", self.epochs, " ", self.verbose)
        self.model.fit(self.X, self.Y_train, epochs=self.epochs, verbose=self.verbose)

    def predecir(self, compuerta):
        # Evaluar predicciones redondeadas
        predicciones = self.model.predict(self.X)
        #print(f"Predicciones para {self.compuerta}:")
        #print(np.round(predicciones))
        return np.round(predicciones)



# Funciones Principales y Comunes de activacion
# 'relu' : Rectified Linear Unit. Modifica valores negativos a cero.
# Es el estándar para capas ocultas.
# 'softmax': Curva en forma de S. Escala los valores entre 0 y 1.
# Ideal para clasificación binaria.
# 'softmax': Convierte salidas en probabilidades que suman 1.
# Obligatorio para clasificación multiclase.
# 'tanh': Tangente hiperbólica. Escala valores entre -1 y 1.
# Utilizada en redes recurrentes.
# 'linear': Activación lineal (identidad). No aplica cambios.
# Es el valor por defecto si no especificas ninguno.

# Funciones Avanzadas y Modernas
# 'gelu': Gaussian Error Linear Unit.
# Usada en modelos tipo Transformer (como BERT o GPT).
# 'swish': Descubierta por Google.
# Suele superar a ReLU en redes muy profundas.
# 'silu': softmax Linear Unit.
# Es exactamente la función Swish con parámetro beta = 1.
# 'mish': Función auto-regularizada que permite un flujo de gradiente más suave.
# 'selu': Scaled Exponential Linear Unit.
# Permite la auto-normalización de la red (mantiene media 0 y varianza 1).
# 'elu': Exponential Linear Unit.
# Reduce el problema del "gradiente moribundo" permitiendo valores negativos suaves.

# Funciones Menos Comunes o Especializadas
# 'softplus': Versión suave y derivable de la función ReLU.
# 'softsign': Alternativa a la tangente hiperbólica con una caída más suave.
# 'exponential': Función exponencial pura.
# Se debe usar con cuidado por riesgo de explosión de gradientes.
# 'hard_softmax': Aproximación computacionalmente más rápida de la función softmaxe.

# Capas de Activación Avanzadas (Clases)
# Si necesitas modificar los parámetros internos de la función (como la pendiente en
# los negativos), no se usan como un simple string, sino como capas independientes
# (keras.layers):
# LeakyReLU: Permite un pequeño flujo de gradiente cuando la unidad está inactiva.
# PReLU: Parametric ReLU. Hace que la pendiente de los negativos sea un parámetro que
# la red aprende.
# ThresholdedReLU: Activa la neurona solo si supera un umbral específico.





```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
# ==========================================
# 1. ARQUITECTURA DE UNA SOLA NEURONA
# ==========================================
class Perceptron(nn.Module): # usamos herencia
    def __init__(self):
        super(Perceptron, self).__init__() # llamando al padre
        # Una sola neuronaa emplear: 2 entradas, 1 salida
        self.linear = nn.Linear(4, 3)
        # Funcion de activacion
        #self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        return self.linear(x)


class RedNeuronal:
    # ==========================================
    # 2. CONFIGURACIÓN DE DATOS
    # ==========================================
    X = torch.tensor(IRIS, dtype=torch.float32)
    Y = torch.tensor(CLASS_IRIS, dtype=torch.long)

    # ==========================================
    # 3. ENTRENAMIENTO Y EVALUACIÓN
    # ==========================================
    # Se usan exactamente los mismos parametros para las tres compuertas

    def ejecutarRed(self, epochs, learning_rate):

      print(f"El entranaiento se hizo con {epochs} epocas con una tasa de aprendizeje de {learning_rate}")
      model = Perceptron()                # instancia a la clase
      criterion = nn.CrossEntropyLoss()
      optimizer = optim.SGD(model.parameters(), lr=learning_rate)

      # Ciclo de entrenamiento
      for epoch in range(epochs):    # para cada epoca
          outputs = model(self.X)         # tomamos el vector
          loss = criterion(outputs, self.Y)    # tomamos los criterios para la perdida

          optimizer.zero_grad()           # optimizador del gradiente cero
          loss.backward()                 # calcula la perdida
          optimizer.step()                # optimiza por paso

      # Evaluacion y diagnostico del aprendizaje si hay fallo
      print(f"Resultado Final:")
      with torch.no_grad():
          predicciones = model(self.X)                            # vuelve a tomar el vector
          predicciones = torch.argmax(predicciones,dim=1)       # hace la prediccion
          # Diagnostic prints to understand the ValueError
          print(f"DEBUG: Type of 'predicciones' before return: {type(predicciones)}")
          print(f"DEBUG: Type of 'model' before return: {type(model)}")
          return predicciones, model # Return both predictions and the trained model
      """
          for i in range(len(self.X)):
              entrada = self.X[i].tolist()                        # la entrada
              pred_int = predicciones[i].item()
              real_int = self.Y[i].item()
              probabilidad = predicciones[i].item()               # toma la probabilidad de prediccion

              # Marcamos con un emoji si el perceptron acerto o fallo
              status = "✅" if pred_int == real_int else "❌ ¡FALLO!"

              print(f"  Entrada: {entrada} -> Predicho: {pred_int} | Real: {real_int} | Prob: {probabilidad:.4f} {status}")
"""
```
